```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.random.randn(10000)
plt.hist(x, bins=100, color='cyan', density=False, alpha=0.8)
plt.title('Histogram')
plt.xlabel('Value')
plt.ylabel('Sum of element')
plt.show()
```

```python
# Multiple Boxplot
x = np.random.randn(10000)*10+3
y = np.random.randn(10000)*10+5
z = np.random.randn(10000)*10+1
plt.boxplot([x, y, z])
plt.title('Multiple Boxplot')
plt.xlabel('Value')
plt.ylabel('Sum of element')
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
z = np.cos(x)

fig0 = plt.figure(figsize=(5,2), dpi =100)
axes0 = fig0.add_axes([0,0,1,1])
axes0.plot(x, y, color='orange', linestyle= '--', label='Sin')
axes0.plot(x, z, color='purple', linestyle= '--', label='Cos')
axes0.legend(loc=0)
axes0.set_xlabel('x')
axes0.set_ylabel('y')
axes0.set_title('Line Plot')
plt.show()
```
