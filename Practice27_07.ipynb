## 1. Transforming DataFrames:

```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv('/content/drive/My Drive/Slides/Examples/Chapter 3/data_studentlist.csv', header='infer')
```

```python
df.colums = ['NAME', 'GENDER', 'GRADE', 'ABSENCES', 'BLOODTYPE', 'HEIGHT', 'WEIGHT']
df.head()
```

### 1.1 Merging DataFrames

```python
df_left = df.loc[:, ['name', 'age', 'gender', 'grade', 'absence']]
df_left_small = df_left.loc[:10, ]
df_left_small
```

```python
df_right = df.loc[:, ['name', 'bloodtype', 'height', 'weight']]
df_right_small = df_right.loc[7:, ]
df_right_small
```

Inner join

```python
pd.merge(df_left_small, df_right_small, on='name', how='inner')
```

Left join

```python
pd.merge(df_left_small, df_right_small,left_on='name', right_on='name', how='left')
```

Right Join

```python
pd.merge(df_left_small, df_right_small, left_on='name', right_on='name', how='right')
```

Full outer join:

```python
pd.merge(df_left_small, df_right_small,left_on='name', right_on='name', how='outer')
```

### 1.2 Binding DataFrames

Bind vertically matched by 'column':

```python
pd.concat([df_left_small, df_right_small], sort= True)
```

Bind horizontally matched by 'index'

```python
pd.concat([df_left_small, df_right_small], axis=1, sort= True)
```

### 2. Data Fearture Engineering:

#### 2.1 Add Columns:

```python
df = pd.read_csv('/content/drive/My Drive/Slides/Examples/Chapter 3/TV_Sales.csv', header='infer')
df.head()
```

```python
df['price_diff'] = df['price'] - df['comp_price']
df.head()
```

```python
df['price_tot'] = df['price'] * df['sales']
df.head( )
```

#### 2.2 Data Type Conversion

```python
df.dtypes
```

```python
df['price_tot'] = df['price_tot'].astype(int)
df.dtypes
```

#### 2.3. Categorical Data Conversion:

```python
count, bin_dividers = np.histogram(df['price'], bins=3)
print(count)
print(bin_dividers)
```

```python
bin_names = ['Low Sales', 'Normal Sales', 'High Sales']

df['sales_bin'] = pd.cut(x=df['sales'],
                     bins=bin_dividers,
                     labels=bin_names,
                     include_lowest=True)

df.head()
```

#### 2.4 One-Hot Vector:

```python
sales_dummies = pd.get_dummies(df['sales_bin']).astype('int')
sales_dummies.head()
```

#### 2.5 Normalization

```python
df['sales'].describe()  #std: độ lệnh chuẩn, mean: trung bình cộng
```

```python
df['sales'] = (df['sales'] - df['sales'].min()) / (df['sales'].max() - df['sales'].min())
df['sales'].describe()
```

```python
cols = ['comp_price', 'income', 'price', 'price_diff', 'price_tot']
df[cols].describe()
```
