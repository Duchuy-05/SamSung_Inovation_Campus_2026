1. Matplotlib visualization with objects

```python
import matplotlib.pyplot as plt
import numpy as np
```

1.1 Visualization with figure object

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
z = np.cos(x)
```

```python
fig0= plt.figure()
axes0 = fig0.add_axes([0, 0, 1, 1]) #tọa độ left, bottom, width, heigh
axes0.plot(x, y, color='red', linestyle='-')  #màu đỏ, nét liền
axes0.set_xlabel('x') #trục tung: vertical axis
axes0.set_ylabel('y') #trục hoành: horizontal axis
axes0.set_title('LINE PLOT')
plt.show()
```

```python
fig0=plt.figure()
axes0 = fig0.add_axes([0, 0, 1, 1])             # Left, bottom, width, height.
axes0.plot(x, y, color='purple', linestyle='-')
axes0.set_xlabel('X')
axes0.set_ylabel('Y')
axes0.set_title('LINE PLOT')
axes0.set_xlim([0, 10])                        # Horizontal axis limits.
axes0.set_ylim([-2, 2])                       # Vertical axis limits.
plt.show()
```

Specify:chỉ định the width and heigh with 'figsize' argument:đối số

```python
fig0=plt.figure(figsize=(10, 1))    # Width and height specified with figsize.
axes0 = fig0.add_axes([0, 0, 1, 1]) # Left, bottom, width, height.
axes0.plot(x, y, color='red', linestyle='-')
axes0.set_xlabel('X')
axes0.set_ylabel('Y')
axes0.set_title('LINE PLOT')
plt.show()
```

1.2. Multiple plots within the same axes:

```python
fig0=plt.figure(figsize=(5,2), dpi=100)
axes0 = fig0.add_axes([0, 0, 1, 1])
axes0.plot(x, y, color='cyan', linestyle='-')
axes0.plot(x, z, color='red', linestyle='--')
axes0.set_xlabel('X')
axes0.set_ylabel('Y')
```

1.4. Multiple plots in an array of axes:

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(6, 5))
#[0,0]
c= np.random.randn(10000)
axes[0,0].hist(c, bins=100, color='red', density=True, alpha=0.8)
axes[0,0].set_title('Histogram')
#[1,0]
axes[1,0].plot(x, y, color='red', linestyle='none', marker='^', markersize=10, markerfacecolor='purple', markeredgecolor='red', alpha=0.8)
axes[1,0].set_xlabel('X')
axes[1,0].set_ylabel('Y')
axes[1,0].set_title('COSINE')
#[0,1]
a = np.array(['Q1', 'Q2', 'Q3', 'Q4'])
b = np.array([234.0, 254.7, 144.6, 317.6])
axes[0,1].bar(a, b, color='green', alpha=0.5)
axes[0,1].set_title('BAR PLOT')
#[1,1]
axes[1,1].plot(x, y, color='orange', linestyle='-')
axes[1,1].set_xlabel('X')
axes[1,1].set_ylabel('Y')
axes[1,1].set_title('SINE')

plt.tight_layout()
plt.show()
```
