### 1. Discrete probability distribution: phân phối xác suất rời rạc

```python
import scipy.stats as st
import matplotlib.pyplot as plt
import numpy as np
```

#### 1.1 Nhị thức

```python
n= 10
p= 0.5
```

```python
# Probability distribution với x = 5
# PMF: function Mass Probability
x = st.binom.pmf(5, n, p)  #Formula: P(X=5)=(n/5)p^5 x (1−p)^(n−5)
# 5: số lần "thành công" (k) để tính xác suất = 5
# n: tổng số phép thử (trials) = 10
# p: xác suất thành công trong mỗi phép thử = 0.5
print('Số khả năng bạn nhận được 5 mặt ngửa trong 10 lần tung đồng xu là: ', x*100, '%')
```

```python
# Quantile at alpha = 0.3
# PPF: function Percent Point
y = st.binom.ppf(0.3, n, p)  # Formula: tìm k nhỏ nhất sao cho P(X ≤ k) ≥ alpha
# 0.3: mức phân vị (alpha) cần tìm = 0.3 (tức 30%)
# n: tổng số phép thử (trials) = 10
# p: xác suất thành công trong mỗi phép thử = 0.5
print('Giá trị k nhỏ nhất để P(X ≤ k) ≥ 30% là: ', y)
```

```python
# P(0 <= x <= 5)
# CDF: Function Distribution Cumulative:tích lũy
z = st.binom.cdf(5, n, p)  # Formula: P(X≤5) = Σ (n/k) p^k x (1−p)^(n−k), k chạy từ 0 đến 5
# 5: giá trị k tối đa cần tính xác suất tích lũy đến
print('Xác suất để có từ 0 đến 5 mặt ngửa trong 10 lần tung đồng xu là: ', z*100, '%')
```

```python
# P(3 <= x <= 7)
w = st.binom.cdf(7, n, p) - st.binom.cdf(2, n, p)  # Formula: P(3≤X≤7) = P(X≤7) - P(X≤2)
# 7: cận trên của khoảng cần tính
# 2: cận dưới trừ 1
# n: tổng số phép thử (trials) = 10
# p: xác suất thành công trong mỗi phép thử = 0.5
print('Xác suất để có từ 3 đến 7 mặt ngửa trong 10 lần tung đồng xu là: ', w*100, '%')
```

```python
# Visualizating the probability distribution
x = np.arange(0, 11)
plt.scatter(x, st.binom.pmf(x, n, p), color='red')
plt.show()
```

### 1.2 Poisson:

```python
# The lambda paramater
lamb = 2
```

```python
# Probability distribution at x = 2
v = st.poisson.pmf(2, lamb)  # Formula: P(X=2) = (e^(-lamb) * lamb^2) / 2!
# 2: số lần xảy ra sự kiện (k) cần tính xác suất
# lamb: số lần xảy ra trung bình (mean rate)
print('Xác suất để sự kiện xảy ra đúng 2 lần là: ', v*100, '%')
```

```python
# Quantile at alpha = 0.5
u = st.poisson.ppf(0.5, lamb)  # Formula: tìm k nhỏ nhất sao cho P(X ≤ k) ≥ alpha
# 0.5: mức phân vị (alpha) cần tìm = 0.5 (tương đương median)
# lamb: số lần xảy ra trung bình (mean rate)
print('Giá trị k nhỏ nhất để P(X ≤ k) ≥ 50% là: ', u)
```

```python
# P(0 <= x <= 5)
t = st.poisson.cdf(5, lamb)  # Formula: P(X≤5) = Σ (e^(-lamb) * lamb^k / k!), k chạy từ 0 đến 5
# 5: giá trị k tối đa cần tính xác suất tích lũy đến
# lamb: số lần xảy ra trung bình (mean rate)
print('Xác suất để sự kiện xảy ra từ 0 đến 5 lần là: ', t*100, '%')
```

```python
# P(3 <= x <= 7)
s = st.poisson.cdf(7, lamb) - st.poisson.cdf(2, lamb)  # Formula: P(3≤X≤7) = P(X≤7) - P(X≤2)
# 7: cận trên của khoảng cần tính
# 2: cận dưới trừ 1
print('Xác suất để sự kiện xảy ra từ 3 đến 7 lần là: ', s*100, '%')
```

```python
# Visualizing the probability distribution.
x=np.arange(0, 11)
plt.scatter(x, st.poisson.pmf(x, lamb), color='green')
plt.show()
```

### 1. Continuous probability density:

#### 1.1. Uniform:

```python
# The a and b parameters.
a = 2000
b = 5000
base = b - a
```

```python
# Probability density: tỉ trọng, mật độ at x = 4500
r = st.uniform.pdf(4500, a, base)  # Formula: f(x) = 1 / (b - a), với a ≤ x ≤ b
# 4500: giá trị x cần tính mật độ xác suất
# a: cận dưới của khoảng (loc)
# base: độ rộng khoảng = b - a (scale), KHÔNG phải cận trên b
print('Mật độ xác suất tại x = 4500 là: ', r)
```

```python
# Quantile at alpha = 0.5
q1 = st.uniform.ppf(0.5, a, base)  # Formula: x = a + alpha * base
# 0.5: mức phân vị (alpha) cần tìm = 0.5
# a: cận dưới của khoảng
# base: độ rộng khoảng = b - a
print('Giá trị x tại phân vị 50% là: ', q1)
```

```python
# Quantile:phân vị at alpha = 0.01
q2 = st.uniform.ppf(0.01, a, base)  # Formula: x = a + alpha * base
# 0.01: mức phân vị (alpha) cần tìm = 0.01
# a: cận dưới của khoảng
# base: độ rộng khoảng = b - a
print('Giá trị x tại phân vị 1% là: ', q2)
```

```python
# Quantile at alpha = 0.99
q2 = st.uniform.ppf(0.99, a, base)  # Formula: x = a + alpha * base
# 0.01: mức phân vị (alpha) cần tìm = 0.01 (tức 1%, gần sát cận dưới)
# a: cận dưới của khoảng
# base: độ rộng khoảng = b - a
print('Giá trị x tại phân vị 99% là: ', q2)
```

```python
# P(a <= x <= 2500)
p1 = st.uniform.cdf(2500, a, base)  # Formula: P(X≤2500) = (2500 - a) / base
# 2500: giá trị x cần tính xác suất tích lũy đến
# a: cận dưới của khoảng
# base: độ rộng khoảng = b - a
print('Xác suất để x nằm từ a đến 2500 là: ', p1*100, '%')
```

```python
# P(2500 <= x <= 3000)
p2 = st.uniform.cdf(3000, a, base) - st.uniform.cdf(2500, a, base)  # Formula: P(2500≤X≤3000) = P(X≤3000) - P(X≤2500)
# 3000: cận trên của khoảng cần tính
# 2500: cận dưới của khoảng cần tính
# a: cận dưới của toàn bộ khoảng phân phối
# base: độ rộng khoảng = b - a
print('Xác suất để x nằm từ 2500 đến 3000 là: ', p2*100, '%')
```

```python
# Visualizing the probability density.
x=np.linspace(1000, 6000, 1000)
plt.plot(x, st.uniform.pdf(x, a, base), color='cyan')
plt.show()
```

#### 1.2. Normal:

```python
# The mu and sigma parameters.
mu = 5
sigma = 2
```

```python
# Probability density:mật độ at x = 0.
st.norm.pdf(0, loc=mu, scale=sigma)
```

```python
# Quantile at alpha = 0.5
st.norm.ppf(0.5, loc=mu, scale=sigma)
```

```python
# Quantile at alpha = 0.01
st.norm.ppf(0.01, loc=mu, scale=sigma)
```

```python
# Quantile at alpha = 0.99
st.norm.ppf(0.99, loc=mu, scale=sigma)
```

```python
# P( x <= 5)
st.norm.cdf(5, loc=mu, scale=sigma)
```

```python
# P( 3 <= x <= 7)
st.norm.cdf(7, loc=mu, scale=sigma) - st.norm.cdf(3, loc=mu, scale=sigma)
```

```python
# P( -1 <= x <= 11)
st.norm.cdf(11, loc=mu, scale=sigma) - st.norm.cdf(-1, loc=mu, scale=sigma)
```

```python
# The location parameter mu.
x=np.linspace(-10, +10, 1000)
sigma=2
mu_1 = -1
mu_2 = 0
mu_3 = +1
plt.plot(x, st.norm.pdf(x, loc=mu_1, scale=sigma), color='red')
plt.plot(x, st.norm.pdf(x, loc=mu_2, scale=sigma), color='yellow')
plt.plot(x, st.norm.pdf(x, loc=mu_3, scale=sigma), color='green')
plt.show()
```

#### 1.3. Chi-square:

```python
# The degree of freedom.
df = 5
```

```python
# Probability density at x = 1
n1 = st.chi2.pdf(1, df)  # Formula: f(x) = 1/(2^(k/2)*Gamma(k/2)) * x^(k/2-1) * e^(-x/2)
# 1: giá trị x cần tính mật độ xác suất
# df: số bậc tự do (degrees of freedom)
print('Mật độ xác suất tại x = 1 là: ', n1)
```

```python
# Quantile at alpha = 0.5
m1 = st.chi2.ppf(0.5, df)  # Formula: tìm x sao cho P(X ≤ x) = 0.5 (median)
# 0.5: mức phân vị (alpha) cần tìm = 0.5
print('Giá trị x tại phân vị 50% là: ', m1)
```

```python
# Quantile at alpha = 0.01
m2 = st.chi2.ppf(0.01, df)  # Formula: tìm x sao cho P(X ≤ x) = 0.01
# 0.01: mức phân vị (alpha) cần tìm = 0.01
print('Giá trị x tại phân vị 1% là: ', m2)
```

```python
# Quantile at alpha = 0.99
m3 = st.chi2.ppf(0.99, df)  # Formula: tìm x sao cho P(X ≤ x) = 0.99
# 0.99: mức phân vị (alpha) cần tìm = 0.99
print('Giá trị x tại phân vị 99% là: ', m3)
```

```python
# P(x <= 5) with df=10
c1 = st.chi2.cdf(5, df=10)  # Formula: P(X≤5) với bậc tự do df=10
# 5: giá trị x cần tính xác suất tích lũy đến
# df=10: số bậc tự do
print('Xác suất để x <= 5 (với df=10) là: ', c1*100, '%')
```

```python
# P(5 <= x <= 10) with df=5
c2 = st.chi2.cdf(10, df=5) - st.chi2.cdf(5, df=5)  # Formula: P(5≤X≤10) = P(X≤10) - P(X≤5), cùng df=5
# 10: cận trên của khoảng cần tính
# 5: cận dưới của khoảng cần tính (lấy trực tiếp, không trừ 1 vì phân phối liên tục)
# df=5: số bậc tự do (phải giống nhau cho cả 2 lần gọi cdf)
print('Xác suất để x nằm từ 5 đến 10 (với df=5) là: ', c2*100, '%')
```

```python
# Visualizing the probability density with different degrees of freedom (df).
x=np.linspace(0, +15, 1000)
plt.plot(x, st.chi2.pdf(x, df=1), color='red')         # df = 1
plt.plot(x, st.chi2.pdf(x, df=5), color='green')       # df = 5
plt.plot(x, st.chi2.pdf(x, df=10), color='blue')       # df = 10
plt.ylim([0, 1])
plt.show()
```
