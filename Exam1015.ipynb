#### Các phép toán trong Series

```python
ser1 = pd.Series(data = [10,20,30,40,5], index=['a', 'b', 'c', 'd', 'e'])
ser2 = pd.Series(data = [10,25,35,40,50], index=['e', 'd', 'c', 'b', 'a'])
```

```python
ser1 + ser2
```

```python
ser1.sum()
```

```python
ser1.min()
```

```python
ser1.mean() # giá trị trung bình
```

```python
ser1.median() # giá trị trung vị
```

```python
ser1.max()
```

```python
ser1.std() # độ lệch chuẩn
```

```python
ser1.var() # Phương sai
```

#### Sử dụng hàm lamba trong Series

```python
ser_heigh = pd.Series([150, 170, 180], name='heigh')
```

```python
plus_10 = lambda x: x + 10
plus_10(200)
```

```python
ser= ser_heigh.apply(lambda x: x + 20)
ser
```
