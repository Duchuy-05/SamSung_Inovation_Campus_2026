```python
import sympy as sp
# x^2020 + 3*x^3-1
x = sp.Symbol('x')
fx= x**2020 + 3*x**3-1
fx
```

```python
sp.diff(fx, x)
```

```python
gx = x/(x+1)
gx
```

```python
print("g'(x)= ",sp.simplify(sp.diff(gx,x)))
```

```python
fx = sp.exp(x)/x
sp.diff(fx,x)
```

```python
gx = 3*sp.log(x,2) - x*sp.log(x)
gx
print("g'(x)= ", sp.simplify(sp.diff(gx, x)))
```
