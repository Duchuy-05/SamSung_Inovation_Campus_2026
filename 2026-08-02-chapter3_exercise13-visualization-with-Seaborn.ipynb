```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
import warnings
```

Read in the data set 'mpg' from the library:

```python
dat = sns.load_dataset('mpg')
```

```python
dat.shape
```

```python
dat.head(5)
```

#### 1.1. Histogram:

```python
sns.displot(dat.mpg, kde=False, rug=False, bins=50, color='red')
plt.show()  #kde: vẽ đường cong KDE
```

```python
sns.kdeplot(dat.mpg, color='blue')  #sns.kdeplot(): function để vẽ đường cong KDE
plt.show()
```

#### 1.2. Scatter plot:

```python
sns.jointplot(x='weight', y='mpg', data=dat, color='red', kind='scatter')
plt.show()
```

```python
sns.jointplot(x='weight', y='mpg', data=dat, color='red', kind='reg')
plt.show()  #reg: bao gồm cột, đường KDE, phân tán/điểm
```

```python
sns.lmplot(data=dat, x='weight', y='mpg')  #sns.lmplot(): chỉ vẽ đường và phân tán/điểm
plt.show()
```

```python
sns.lmplot(data=dat, x='weight', y='mpg', col='origin', hue='origin')
plt.show() #chia dữ liệu thành các đồ thị riêng biệt
```

#### 1.3. Heat map:

```python
dat = sns.load_dataset('mpg')
```

```python
dat.drop(['origin', 'name'], axis=1, inplace=True)
```

```python
x = dat.corr()
x
```

```python
sns.heatmap(x)
plt.show()
```

```python
# Heatmap.
# Adjust the center.
sns.heatmap(x, cmap='coolwarm', center=-1)
plt.show()
```

#### 1.4. PairGrid:

```python
g=sns.PairGrid(dat)
g.map(plt.scatter)
plt.show()
```

```python
g=sns.PairGrid(dat)
g.map_diag(sns.displot, color='orange')         # Diagonal = histogram.
g.map_upper(plt.scatter, color='green')          # Upper triangle = scatter plot.
g.map_lower(sns.kdeplot, color='blue')           # Lower triangle = KDE.
plt.show()
```

#### 1.5. Pairplot

```python
dat = sns.load_dataset('iris') #lấy dữ liệu csv từ drive
```

```python
sns.pairplot(dat)
plt.show()
```

```python
sns.pairplot(dat, hue='species')
plt.show()
```

```python
sns.pairplot(dat, hue='species', palette='coolwarm')
plt.show()
```

#### 1.6. FacetGrid:

```python
g=sns.FacetGrid(data=dat, col='species')
g.map(sns.displot, 'petal_length')
plt.show()
```

```python
g=sns.FacetGrid(data=dat, row='species')
g.map(sns.displot, 'petal_length')
plt.show()
```

```python
g=sns.FacetGrid(data=dat, col='species')
g.map(plt.scatter, 'petal_length', 'petal_width', color='orange')
plt.show()
```
